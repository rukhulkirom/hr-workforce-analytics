# Range Validation & Outlier Assessment

Range validation is performed to identify values that may violate
expected business or logical constraints.

Statistical outliers are not automatically treated as data errors.
Extreme values will be evaluated based on their business context.

```python
expected_ranges = {
    "Age": (18, 60),
    "DistanceFromHome": (1, None),
    "Education": (1, 5),
    "EnvironmentSatisfaction": (1, 4),
    "JobInvolvement": (1, 4),
    "JobLevel": (1, 5),
    "JobSatisfaction": (1, 4),
    "PerformanceRating": (3, 4),
    "RelationshipSatisfaction": (1, 4),
    "StockOptionLevel": (0, 3),
    "WorkLifeBalance": (1, 4),
    "TrainingTimesLastYear": (0, None),
    "NumCompaniesWorked": (0, None),
    "TotalWorkingYears": (0, None),
    "YearsAtCompany": (0, None),
    "YearsInCurrentRole": (0, None),
    "YearsSinceLastPromotion": (0, None),
    "YearsWithCurrManager": (0, None),
    "PercentSalaryHike": (0, None),
}
```

```python
range_results = []

for column, (min_expected, max_expected) in expected_ranges.items():
    actual_min = df[column].min()
    actual_max = df[column].max()

    below_min = (
        actual_min < min_expected
        if min_expected is not None
        else False
    )

    above_max = (
        actual_max > max_expected
        if max_expected is not None
        else False
    )

    range_results.append({
        "column": column,
        "actual_min": actual_min,
        "actual_max": actual_max,
        "expected_min": min_expected,
        "expected_max": max_expected,
        "below_min": below_min,
        "above_max": above_max,
        "valid_range": not (below_min or above_max)
    })

range_results = pd.DataFrame(range_results)

range_results
```

```python
numeric_columns = df.select_dtypes(include="number").columns

negative_values = {
    column: int((df[column] < 0).sum())
    for column in numeric_columns
}

negative_values = {
    column: count
    for column, count in negative_values.items()
    if count > 0
}

negative_values
```

```python
outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": int(outlier_count),
        "outlier_percentage": round(
            outlier_count / len(df) * 100,
            2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    "outlier_count",
    ascending=False
)
```

## Outlier Assessment Conclusion

Statistical outlier detection using the IQR method identified several
extreme observations across variables such as `TrainingTimesLastYear`,
`PerformanceRating`, `MonthlyIncome`, and career tenure variables.

However, these observations were not automatically removed.

Business-rule validation confirmed that the identified extreme values
fall within valid business ranges and do not violate logical relationships
between workforce variables.

For example:

- `PerformanceRating = 4` is a valid rating category even though it is
  statistically flagged because the IQR is zero.
- Higher `MonthlyIncome` values can be valid for employees with higher
  job levels.
- Higher tenure values represent valid long-term employees.
- Higher `TrainingTimesLastYear` values represent valid training activity.

Therefore, no observations will be removed solely based on statistical
outlier detection.

Extreme values will be retained for analysis unless a specific business
or data-quality rule identifies them as invalid.
